```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv("raw_fuel_prices_messy.csv") # Load csv file


print(df.head()) # prints first five lines

print(df.info()) # Checking missing values and datatypes at the same time.

print(df.isnull().sum()) # Checking missing values in each column 

print(df.describe()) # Prints basic statistics

```

```python
# Drop row with missing date
df.dropna(subset=['date'], inplace=True)

# Calculate averages
avg_petrol = df['petrol_95'].mean()
avg_diesel = df['diesel'].mean()

# Fill missing values (correct pattern)
df['petrol_95'] = df['petrol_95'].fillna(avg_petrol)
df['diesel'] = df['diesel'].fillna(avg_diesel)

# Convert date
df['date'] = pd.to_datetime(df['date'])

# Verify
print(df.info())
print("\nMissing values:")
print(df.isnull().sum())
```

```python
conn = sqlite3.connect("fuel_prices_messy.db")

df_check = pd.read_sql("SELECT * FROM fuel_prices_clean", conn)
print(df_check.head())
print(f"\nTotal rows: {len(df_check)}")

conn.close()
```

```python
conn = sqlite3.connect("fuel_prices_messy.db")

# Checking for duplicates

query = """
SELECT date, COUNT(*) as count
FROM fuel_prices_clean
GROUP BY date
HAVING COUNT(*) > 1
"""

duplicates = pd.read_sql(query,conn)

print(duplicates)
conn.close()
```

```python
conn = sqlite3.connect("fuel_prices_messy.db")

# Count total rows
query = """
SELECT COUNT(*) as total_rows
FROM fuel_prices_clean
"""
count = pd.read_sql(query,conn)
print("  Total Rows  ")
print(count)

conn.close()
```
